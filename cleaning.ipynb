```python
import pandas as pd

df = pd.read_csv("../dataset/college_student_career_preference_dataset.csv")

print("Dataset loaded successfully!")
```

```python
print("Dataset Shape:")
print(df.shape)
```

```python
print("Column Names:")
print(df.columns.tolist())
```

```python
print("Data Types:")
print(df.dtypes)
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
print("Career Preferences:")
print(df["Career_Preference"].value_counts())
```

```python
df.to_csv("original_dataset.csv", index=False)

print("Original dataset saved successfully!")
```

```python
cleaning_summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str),
    "Missing Values": df.isnull().sum(),
    "Unique Values": df.nunique()
})

print(cleaning_summary)
```

```python
cleaning_summary.to_csv("data_cleaning_summary.csv", index=False)

print("Data cleaning summary saved successfully!")
```

```python
duplicates = df[df.duplicated()]

print("Number of duplicate rows:", len(duplicates))
```

```python
duplicates.to_csv("duplicate_rows.csv", index=False)

print("Duplicate rows saved successfully!")
```

```python
df_cleaned = df.drop_duplicates()

print("Original shape:", df.shape)
print("Cleaned shape:", df_cleaned.shape)
```

```python
df_cleaned.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
